```python
payroll_number = int(input("Enter payroll number:"))
name = str(input("Enter employee name:"))
gender = str(input("Enter gender:(M/F)"))
department=str(input("Enter department:"))
basic_salary = float(input("Enter basic salary:"))
```

```python
house_allowance = 6500
medical_allowance = 5500
```

```python
gross_pay = basic_salary + house_allowance + medical_allowance
```

```python
if gross_pay>=40000:
 paye=gross_pay*0.06
elif gross_pay>=31000:
 paye=gross_pay*0.05
elif gross_pay>=20000:
 paye=gross_pay*0.04
else:
 paye=0
```

```python
nhif = gross_pay*0.02
nssf = basic_salary*0.03
```

```python
total_deductions=paye+nhif+nssf
net_pay=gross_pay-total_deductions
```

```python
print("\n-----EMPLOYEE PAYSLIP------")
print("Payroll Number:",payroll_number)
print("Name:",name)
print("Gender:",gender)
print("Department:",department)
print("Basic Pay: Ksh.",basic_salary)
print("Gross Pay: Ksh.",gross_pay)
print("P.A.Y.E: Ksh.",paye)
print("N.H.I.F: Ksh.",nhif)
print("N.S.S.F: Ksh.",nssf)
print("Total Deductions:(netpay) Ksh.",net_pay)
```
