```python
name = str(input("Enter patient's name:"))
gender =str(input("Enter gender:M/F "))
age = int(input("Enter age: "))
residence = str(input("Enter place of residence: "))
```

```python
print("\nPossible symptoms Combination:")
print("1.Vomiting and Diarrhea")
print("2.Vomiting and fever")
print("3.Chest pain and wheezing sound")
print("4.Frequent urination and fluctuations of sugar levels")
symptom1 = input("Enter symptom 1:").lower()
symptom2 = input("Enter symptom 2:").lower()
```

```python
if(symptom1 == "vomiting" and symptom2 =="diarrhea") or\
    (symptom1 == "diarrhea"and symptom2 == "vomiting"):
    diagnosis = "Typhoid"
elif(symptom1 == "vomiting" and symptom2 =="fever") or\
    (symptom1 == "fever" and symptom2=="vomiting"):
    diagnosis="Malaria"
elif(symptom1 =="chest pain" and symptom2 == "wheezing sound") or\
    (symptom1 == "wheezing sound" and symptom2 == "chest pain"):
    diagnosis = "Pneumonia"
elif(symptom1 == "frequent urination" and symptom2 == "fluctuations of sugar levels") or\
    (symptom1 == "fluctuations of sugar levels" and symptom2 == "frequent urination"):
    diagnosis = "Diabetes"
else:
    diagnosis="No matching diagnosis found"
```

```python
print("\n-----AMANI MEDICAL DIAGNOSIS REPORT----")
print("Patient Name:",name)
print("Gender:",gender)
print("Age:",age)
print("Residence:",residence)
print("Symptom 1:",symptom1)
print("Symptom 2:",symptom2)
print("Possible Diagnosis:",diagnosis)


```
